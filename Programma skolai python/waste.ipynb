```python
# ==============================
# Laser Waste Calculator
# Manual Input Version
# ==============================

print("Laser Cutting Waste Calculator\n")

# --- USER INPUT ---
sheet_width = float(input("Enter sheet width (mm): "))
sheet_height = float(input("Enter sheet height (mm): "))
used_area = float(input("Enter used area (mm²): "))

# --- CALCULATION ---
sheet_area = sheet_width * sheet_height
waste_area = sheet_area - used_area

if used_area > sheet_area:
    print("\nERROR: Used area cannot be larger than sheet area.")
else:
    waste_percentage = (waste_area / sheet_area) * 100

    # --- OUTPUT ---
    print("\n--- Waste Report ---")
    print(f"Sheet Area: {sheet_area:,.0f} mm²")
    print(f"Used Area:  {used_area:,.0f} mm²")
    print(f"Waste Area: {waste_area:,.0f} mm²")
    print(f"Waste:      {waste_percentage:.2f} %")
```
